```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")

for ent in doc.ents:
    print(ent.text, " | ", ent.label_, " | ", spacy.explain(ent.label_))
```

```python
nlp.pipe_labels['ner']  #List down all the entities
```

Custom NER, a span tells spaCy exactly which part of the sentence is the entity.

```python
text = "I know Python and Java"
doc = nlp(text)

span1 = doc.char_span(7, 13, label="SKILL")
span2 = doc.char_span(18, 22, label="SKILL")

doc.ents = [span1, span2]

for ent in doc.ents:
    print(ent.text, "→", ent.label_)
```
