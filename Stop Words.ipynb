```python
nlp =  spacy.load("en_core_web_sm")
doc = nlp("i am learning Python and I am interested in machine learning.")
for token in doc:
    if token.is_stop:
        print(token)
```

```python
def preprocess(text):
    doc =  nlp(text)
    no_stop_words = [token.text for token in doc if not token.is_stop]
    return (no_stop_words)
```
