#### Task 1: Get all the proper nouns from a given text in a list and count how many.

A proper Noun is a noun that names a particular person, place, or thing.

```python
proper_noun_text = '''
In 2025, Peter traveled to Delhi to attend a technology summit. There, he met Loki and discussed ideas for his startup.
Later, Peter visited famous landmarks like the Red Fort and India-Gate. Meanwhile, Satya and Naveen were exploring Jaipur and its beautiful palaces.
After returning to Hyderabad, Peter shared his experiences with his friend Manjunath, who was planning a trip to Chennai.
'''
```

```python
import spacy 
nlp =  spacy.load("en_core_web_sm")
doc = nlp(proper_noun_text)
proper_noun = []
for ent in doc.ents:
    if ent.label_ in ["PERSON", "GPE", "ORG", "LOC", "FAC"]:
        proper_noun.append(ent.text)
print(proper_noun)
print("Count:", len(proper_noun))
```

#### Task 2: Get all companies names from a given text and also the count of them.

- **Hint**: Use the spacy **ner** functionality 

```python
companies_text = '''The Top 5 companies in USA are Tesla, Walmart, Amazon, Microsoft, Google and the top 5 companies in 
India are Infosys, Reliance, HDFC Bank, Hindustan Unilever and Bharti Airtel'''
```

```python
doc = nlp(companies_text)

companies = []

for ent in doc.ents:
    if ent.label_ == "ORG":
        companies.append(ent.text)

print(companies)
print("Count:", len(companies))
```
